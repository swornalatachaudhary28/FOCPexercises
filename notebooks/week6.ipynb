```python
#Write a function that accepts a positive integer as a parameter and then returns a representation of that number in binary (base 2).
def decimal_to_binary(n):
    if n<=0:
        return "Input a positive integer."
    return bin(n)[2:]
n= int(input("enter an interger"))
print (decimal_to_binary(n))
```

```python
#Write and test a function that takes an integer as its parameter and returns the factors of that integer.
def factors(n):
    if n == 0:
        return "0 has infinite factors"
    n=abs(n)
    result=[]
    for i in range(1, n+1):
      if n % i == 0:
        result.append(i)
    return result
n=int(input("enter an integer"))
print(factors(n))
```

```python
#Write and test a function that determines if a given integer is a prime number.
def prime_numbers(n):
    if n <= 1:
        return False
    for i in range (2,n):
        if n % i ==0:
            return False
    return True
n=int(input("Enter a number:"))
if prime_numbers(n):
    print("is Prime Number")
else:
    print("Not a prime Number")

        
        
```
