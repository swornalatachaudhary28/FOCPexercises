```python
#in the range of 0 to 100
def in_the_range(value):
    return 0 <= value <=100 
number = int (input("Enter an integer:"))
if in_the_range(number):
    print ("The number is within the range 0-100.")
else:
    print("The number is outside the range 0-100.")
```

```python
#count upper and lowercase letters
def count_case(text):
    upper_count=0
    lower_count=0
    
    for char in text:
        if char.isupper():
            upper_count+=1
        elif char.islower():
            lower_count+=1
    return upper_count,lower_count
string=input("Enter a string:")
upper,lower=count_case(string)
print("Uppercase letters:",upper)
print("Lowercase letters:",lower)
```

```python
#Write and test a function that converts a temperature measured in degrees centigrade into the equivalent in fahrenheit, and another that does the reverse conversion. Test both functions. (Google will find you the formulae).
def celcius_to_fahrenheit(c):
    return (c*9/5)+32
def fahrenheit_to_celcius(f):
    return(f-32)*5/9
c=float(input("enter the temperature in celcius:"))
print(celcius_to_fahrenheit(c))
f=float(input("enter the temperature in fahrenheit:"))
print(fahrenheit_to_celcius(f))

```

```python
#Write a program that reads 6 temperatures (in the same format as before), and displays the maximum, minimum, and mean of the values.
temps_f=[]
for i in range(6):
    temp=input(f"Enter temperature{i+1}:")
    celcius=float(temp[:-1])
    fahrenheit=(celcius*9/5)+32
    temps_f.append(fahrenheit)
print("Maximum:",max(temps_f))
print("Minimum:",min(temps_f))
print("Mean:",sum(temps_f)/len(temps_f))
```

```python
#Modify the previous program so that it can process any numberof values. The input terminates when the user just pressed "Enter" at the prompt rather than entering a value.
temps_f=[]
while True:
    temp=input("Enter the temperature: or Press enter to terminate the program.")
    if temp == "":
        break
    celcius=float(temp[:-1])
    fahrenheit=(celcius*9/5)+32
    temps_f.append(fahrenheit)
if len(temps_f)>0:
    print("Maximum:",max(temps_f))
    print("Minimum:",min(temps_f))
    print("Mean:",sum(temps_f)/len(temps_f))
else:
    print("No temperatures were entered.")
```
